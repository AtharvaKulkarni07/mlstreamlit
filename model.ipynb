Result - Na_to_K has some outlier. We will use IQR method to remove outlier

```python
# Removing outlier from the data using iqr method
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# removing outliers
num_cols = ['Age','Na_to_K']

for col in num_cols:
    df2 = remove_outliers_iqr(df, col)

# Check shape after removing outliers
print("Dataset shape after removing outliers:", df.shape)
```

# Model Training

```python
X = df2.drop(["Drug"], axis=1)
y = df2["Drug"]

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 0,stratify=y)
```

```python
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

Feature scaling using one hot encoding. I am using pd.get_dummies() function to convert categorical data to numeric

```python
X_train = pd.get_dummies(X_train, drop_first=True,dtype=int)
X_test = pd.get_dummies(X_test, drop_first=True,dtype=int)

# Align columns in train/test in case some categories are missing in test
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

print(X_train.shape, X_test.shape)
```

```python
X_test.head()
```

# Classification Algorithm 

RandomForestClassifier

```python
from sklearn.ensemble import RandomForestClassifier
RF = RandomForestClassifier(max_leaf_nodes=30, class_weight='balanced',random_state=42)
y_pred_rf = RF.fit(X_train, y_train).predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
print(classification_report(y_test, y_pred_rf))
```

```python
print(confusion_matrix(y_test, y_pred_rf))
```

```python
RF_accuracy = accuracy_score(y_test, y_pred_rf)
print("Random Forest Classifier Accuracy:", RF_accuracy)
```

Logistic Regression

```python
from sklearn.linear_model import LogisticRegression
LR = LogisticRegression(max_iter=5000,class_weight='balanced',random_state=42)
y_pred_lr = LR.fit(X_train, y_train).predict(X_test)


print(classification_report(y_test, y_pred_lr))
print(confusion_matrix(y_test, y_pred_lr))
LR_accuracy = accuracy_score(y_test, y_pred_lr)
print("Logistic Regression Accuracy:", LR_accuracy)
```

Decision Tree 

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

DT = DecisionTreeClassifier(class_weight='balanced', random_state=42)
y_pred_dt = DT.fit(X_train, y_train).predict(X_test)

print(classification_report(y_test, y_pred_dt))
print(confusion_matrix(y_test, y_pred_dt))
DT_accuracy = accuracy_score(y_test, y_pred_dt)
print("Decision Tree Accuracy:", DT_accuracy)
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

SVM = SVC(class_weight='balanced', probability=True, random_state=42)
y_pred_svm = SVM.fit(X_train, y_train).predict(X_test)

print(classification_report(y_test, y_pred_svm))
print(confusion_matrix(y_test, y_pred_svm))
SVM_accuracy = accuracy_score(y_test, y_pred_svm)
print("SVM Accuracy:", SVM_accuracy)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

KNN = KNeighborsClassifier(n_neighbors=3)
y_pred_knn = KNN.fit(X_train, y_train).predict(X_test)

print(classification_report(y_test, y_pred_knn))
print(confusion_matrix(y_test, y_pred_knn))
KNN_accuracy = accuracy_score(y_test, y_pred_knn)
print("KNN Accuracy:", KNN_accuracy)
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

NB = GaussianNB()
y_pred_nb = NB.fit(X_train, y_train).predict(X_test)

print(classification_report(y_test, y_pred_nb))
print(confusion_matrix(y_test, y_pred_nb))
NB_accuracy = accuracy_score(y_test, y_pred_nb)
print("Naive Bayes Accuracy:", NB_accuracy)
```

```python
from sklearn.ensemble import GradientBoostingClassifier
GB = GradientBoostingClassifier(random_state=42)
y_pred_gb = GB.fit(X_train, y_train).predict(X_test)
print(classification_report(y_test, y_pred_gb))
print(confusion_matrix(y_test, y_pred_gb))
GB_accuracy = accuracy_score(y_test, y_pred_gb)
print("Gradient Boosting Classifier Accuracy:", GB_accuracy)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Store all model accuracies (update names if you used different variable names)
results = {
    "Logistic Regression": LR_accuracy,
    "Decision Tree": DT_accuracy,
    "Random Forest": RF_accuracy,
    "SVM": SVM_accuracy,
    "KNN": KNN_accuracy,
    "Naive Bayes": NB_accuracy,
    "Gradient Boosting": GB_accuracy
}

# Convert to DataFrame
results_df = pd.DataFrame(list(results.items()), columns=["Model", "Accuracy"])
results_df = results_df.sort_values(by="Accuracy", ascending=False).reset_index(drop=True)

# Display comparison table
print("===== Comparative Accuracy Analysis =====")
print(results_df)
```

```python
from sklearn.model_selection import GridSearchCV
param_grid_rf = {
    'n_estimators': [50, 100, 150],
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
}

param_grid_dt = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [None, 5, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'max_features': [None, 'sqrt', 'log2']
}

```

```python
rf_new = RandomForestClassifier(random_state=42, class_weight='balanced')
grid_search_rf = GridSearchCV(
    estimator=rf_new,
    param_grid=param_grid_rf,
    scoring='accuracy',    # or 'f1_macro' if classes are imbalanced
    cv=5,                  # 5-fold cross-validation
    n_jobs=-1,             # use all cores
    verbose=2
)
```

```python
grid_search_rf.fit(X_train, y_train)   

print("Best Hyperparameters:", grid_search_rf.best_params_)
print("Best CV Accuracy:", grid_search_rf.best_score_)
```
